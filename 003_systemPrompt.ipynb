```python
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()
```

```python
# load the client
client = Anthropic()
# choose the model
model = "claude-haiku-4-5-20251001"
# make an empty messages list
messages = []
```

```python
# 3 helper functions
def add_user_message(messages, text):
    user_message = {"role": "user",
    "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant",
    "content": text}
    messages.append(assistant_message)

system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""
def chat(messages, system = None): # we're not allowed to pass a system prompt as None, so we make the if condition you see
    params = {
        'model': model,
        'max_tokens': 1000,
        'messages': messages
    }
    if system:
        params['system'] = system
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
# Answer before I added the system prompt in the chat function, then this is after adding the system prompt.
messages = []
add_user_message(messages, "How do I solve 5x + 3 = 2 for x?")
answer = chat(messages)

answer
```

```python
# response steaming
messages = []
add_user_message(messages, "Give me 3 sentences about the iPhone 15.")
with client.messages.stream(
    model = model,
    max_tokens = 1000,
    messages = messages
) as stream:
    for text in stream.text_stream:
        print(text, end = "")
```
