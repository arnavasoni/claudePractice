```python
# load env variables
from dotenv import load_dotenv, find_dotenv

load_dotenv()
print(find_dotenv())
```

```python
# create API client
from anthropic import Anthropic

model = "claude-haiku-4-5-20251001" # name of the model to be used
client = Anthropic() # creating a client
```

```python
# to find the currently available models
models = client.models.list()
for m in models.data:
    print(m.id)
```

```python
# make a request
message = client.messages.create(
    model = model, # what model are we using
    max_tokens = 1000, # max number of tokens
    messages = [
        {
            "role": "user", # who's speaking?
            "content": "What is quantum computing? Answer in one sentence." # whatcha saying?
        }
    ]
)
```

```python
# display the text of the response
message.content[0].text
```

```python
# make another request; this shows that claude api doesn't store the conversation by default
message = client.messages.create(
    model = model,
    max_tokens = 1000,
    messages = [
        {
            "role": "user",
            "content": "Write another sentence", # totally different conversation
        }
    ]
)
```

```python
message.content[0].text
```

### Adding context & maintaining conversation for the chatbot

```python
# 3 helper functions

# to store what user's saying
def add_user_message(messages, text):
    user_message = {"role": "user",
    "content": text}
    messages.append(user_message)

# to store what assistant's saying
def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant",
    "content": text}
    messages.append(assistant_message)

# no need to call the client again & again
def chat(messages):
    message = client.messages.create(
    model = model,
    max_tokens = 1000,
    messages = messages,
    )
    return message.content[0].text
```

```python
# make a list of messages
messages = []
# add initial user question
add_user_message(messages, "Define quantum computing in one sentence.")
# get result from assistant
answer = chat(messages)
# add assistant response into messages list
add_assistant_message(messages, answer)
# add another user message
add_user_message(messages, "Write another sentence.")
# get result from assistant
answer = chat(messages)
# add answer to messages list
add_assistant_message(messages, answer)
```
